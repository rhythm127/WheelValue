```python
import pandas as pd
import numpy as np

# Load dataset
DATA_PATH = "../data/raw/cardekho_raw.csv"
df = pd.read_csv(DATA_PATH)

print(f"Dataset Shape: {df.shape[0]} rows, {df.shape[1]} columns")
df.head()
```

```python
# Check data types, missing values, and duplicates
inspection_df = pd.DataFrame({
    "Data Type": df.dtypes,
    "Non-Null Count": df.notnull().sum(),
    "Missing Count": df.isnull().sum(),
    "Missing %": (df.isnull().sum() / len(df) * 100).round(2),
    "Unique Values": df.nunique(),
    "Sample Value": df.iloc[0]
})

print(f"Total Exact Duplicate Rows: {df.duplicated().sum()}")
print(f"Year Range: {df['year'].min()} to {df['year'].max()}")
print(f"Selling Price Skewness: {df['selling_price'].skew():.2f}")
inspection_df
```

```python
# Inspect categorical columns
for col in ['fuel', 'seller_type', 'transmission', 'owner']:
    print(f"\n--- {col.upper()} ---")
    print(df[col].value_counts(dropna=False))

# Inspect text-wrapped numeric columns
for col in ['mileage', 'engine', 'max_power']:
    print(f"\nSample values in '{col}':", df[col].dropna().unique()[:5])
```
